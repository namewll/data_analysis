```python
import jieba
```

```python
index=0
clean_score_list=[]
place_score_list=[]
serve_score_list=[]
price_score_list=[]
for i in df.groupby('酒店名称')['合并后评论']:
    lst=[j for j in i[1]]
    bar_len=len(lst)
    one_bar=''.join([j for j in i[1]])
    all_words=[word for word in jieba.cut(one_bar)]
    clean_score=0
    place_score=0
    serve_score=0
    price_score=0
    for w in all_words:
        if w in dimensions_keywords['清洁程度']['positive']:
            clean_score+=1
        if w in dimensions_keywords['清洁程度']['nav']:
            clean_score-=1
    clean_score=clean_score/bar_len
    for w in all_words:
        if w in dimensions_keywords['地理位置']['positive']:
            place_score+=1
        if w in dimensions_keywords['地理位置']['nav']:
            place_score-=1
    place_score=place_score/bar_len
    for w in all_words:
        if w in dimensions_keywords['服务体验']['positive']:
            serve_score+=1
        if w in dimensions_keywords['服务体验']['nav']:
            serve_score-=1
    serve_score=serve_score/bar_len
    for w in all_words:
        if w in dimensions_keywords['性价比']['positive']:
            price_score+=1
        if w in dimensions_keywords['性价比']['nav']:
            price_score-=1
    price_score=price_score/bar_len
    for len_ in range(bar_len):
        clean_score_list.append(clean_score)
        place_score_list.append(place_score)
        serve_score_list.append(serve_score)
        price_score_list.append(price_score)
```

```python
df['clean_score']=pd.Series(data=clean_score_list)
df['place_score']=pd.Series(data=place_score_list)
df['serve_score']=pd.Series(data=serve_score_list)
df['price_score']=pd.Series(data=price_score_list)
```
